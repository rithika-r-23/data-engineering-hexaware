Section C: PySpark DataFrames

```python
from pyspark.sql import functions as F

reg_df = spark.read.table("event_registrations_source")
event_df = spark.read.table("event_master")
reg_df.printSchema()
event_df.printSchema()
```

```python
display(reg_df.select("registration_id", "attendee_name", "tickets", "unit_price"))
```

```python
display(reg_df.filter((F.col("status") == "Confirmed") & (F.col("tickets") > 2)))
```

```python
amount_df = reg_df.withColumn("total_amount", F.col("tickets") * F.col("unit_price"))
display(amount_df)
```

```python
category_df = amount_df.withColumn(
    "booking_category",
    F.when(F.col("total_amount") >= 2000, "High").otherwise("Regular")
)
display(category_df)
```

```python
city_revenue_df = (
    reg_df.filter(F.col("status") == "Confirmed")
    .join(event_df, on="event_id", how="inner")
    .groupBy("city")
    .agg(F.sum(F.col("tickets") * F.col("unit_price")).alias("total_confirmed_revenue"))
    .orderBy(F.col("total_confirmed_revenue").desc())
)
display(city_revenue_df)
```

Section I: Structured Streaming

```python
spark.sql("CREATE VOLUME IF NOT EXISTS rithika_databricks_7405609940652409.default.checkpoints")
ckpt_base = "/Volumes/rithika_databricks_7405609940652409/default/checkpoints"
```

```python
rate_df = spark.readStream.format("rate").option("rowsPerSecond", 5).load()
rate_df.printSchema()
print(rate_df.isStreaming)
```

```python
display(rate_df, checkpointLocation=ckpt_base + "/q47_rate")
```

```python
txn_df = (
    rate_df
    .withColumn("transaction_id", F.concat(F.lit("TXN-"), F.col("value").cast("string")))
    .withColumn("transaction_amount", (F.col("value") % 100) * 10)
)
display(txn_df, checkpointLocation=ckpt_base + "/q48_txn")
```

```python
high_txn_df = txn_df.filter(F.col("transaction_amount") > 500)
display(high_txn_df, checkpointLocation=ckpt_base + "/q49_high")
```

```python
category_stream_df = (
    txn_df
    .withColumn(
        "transaction_category",
        F.when(F.col("transaction_amount") >= 700, "High")
         .when(F.col("transaction_amount") >= 400, "Medium")
         .otherwise("Low")
    )
    .groupBy("transaction_category")
    .count()
)
display(category_stream_df, checkpointLocation=ckpt_base + "/q50_counts")
```
